```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-np.pi, np.pi)
y_sin = np.sin(np.sin(x))
y_cos= np.abs(np.cos(x))

plt.scatter(x,y_sin, label="sin")
plt.scatter(x,y_cos, label="cos")
plt.xlabel("x",size=14)
plt.ylabel("y",size=14)
plt.grid()
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x= np.linspace(-4,4)
y= np.abs(x**2-4)
print(x)
plt.scatter(x,y)
plt.xlabel("x",size=14)
plt.ylabel("y",size=14)
plt.grid()
plt.show()
```

```python
import numpy as np
a= np.array([2,2,-2,-2])
print("L2 놈:")
print(np.linalg.norm(a))  
print("L1 놈:")
print(np.linalg.norm(a,1))
```

```python
import numpy as np
a= np.array([1,-2,2])
b= np.array([2,-2,1])
print("----내적------")
print(np.dot(a,b))
print("------a의 L2 norm-----")
print(np.linalg.norm(a))
print("------b의 L2 norm -----")
print(np.linalg.norm(b))
print("------a의 L1 norm-----")
print(np.linalg.norm(a,1))
print("------b의 L1 norm -----")
print(np.linalg.norm(b,1))
```
