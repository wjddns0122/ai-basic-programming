```python
# 입력 X 출력 O

def my_func2(p, q):
    print("입력 없고 출력 있는 경우")
    k = p ** q
    print("P의 Q승은: ", k)

print("함수 호출 전임")
a = 7
b = 3
k = my_func2(a, b)
print("After Function Called. k = ", k)
```

```python
# 폰노이만
def my_func4(p, q):
    k = p ** q
    m = q ** p
    return k, m

print("함수 호출 전임")
a = 7
b = 3
k, m = my_func4(a, b)
print("after Function called, k & m", k, m)
```

```python
s = 123

def show_number():
    t = 456
    print(s, t)

show_number()
print(s)
```

```python
ss = 123

def show_arg(aa):
    print("local : ", aa)

show_arg(456)
print("global :", ss)
```

```python
bb = 123

print("before bb : ", bb)
def set_global():
    global bb
    bb = 456
    print("global: ", bb)

set_global()
print("global:", bb)
```
