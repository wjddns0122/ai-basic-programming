```python
import numpy as np 

a = np.array([[0, 1, 2, 3],
              [8, 5, 2, 3],
              [3, 5, 7, 4],
              [9, 5, 1, 2]])

b = np.array([[1, 2, 1], [5, 4, 5], [6, 3, 6]])

det = np.linalg.det(a)
print(det)

det_b = np.linalg.det(b)
print(det_b)

a_inv = np.linalg.inv(a)
print(a_inv)

print(np.dot(a, a_inv))
```

```python
import numpy as np 

a = np.array([[1, 2],
              [5, 6]])

det_a = np.linalg.det(a)

if abs(det_a) < 1.0e-9:
    print("역행렬 구할 수 없음 !!")
else:
    a_inv = np.linalg.inv(a)
    print("a의 역행렬은 : \n", a_inv)
    print("a의 역행렬을 검증함 : \n", np.dot(a, a_inv))
    
```

```python
import numpy as np 
import matplotlib.pyplot as plt 

# 화살표를 그리는 함수
def arrow(start, size, color):
    plt.quiver(start[0], start[1], size[0], size[1], angles="xy", scale_units="xy", scale=1, color=color)

# 화살표어ㅣ 시작점
s = np.array([0, 0])

# 벡터
a = np.array([2, 3])

arrow(s, a, color="black")

# 그래프 표시
plt.xlim((-3, 3))
plt.ylim((-3, 3))
plt.xlabel("x", size=14)
plt.xlabel("y", size=14)
plt.grid()
plt.gca().set_aspect("equal")
plt.show()
```

```python
import numpy as bp
import matplotlib.pyplot as plt

#화살표를 그리는 함수
def arrow(start, size, color):
    plt.quiver(start[0], start[1], size[0], size[1], 
               angles="xy", scale_units="xy", scale=1, color=color)

s = np.array([0,0])
a = np.array([2,3])
b = np.array([-3,-1])

e_x= np.array([1,0])
e_y= np.array([0,1])

arrow(s,b, color="red")
arrow(s,a, color="blue")
arrow(s,e_x, color="black")
arrow(s,e_y, color="black")

plt.xlim(-3,3)
plt.ylim(-3,3)
plt.xlabel("x", size=14)
plt.ylabel("y", size=14)
plt.grid()
plt.gca().set_aspect("equal")
```

```python
import numpy as bp
import matplotlib.pyplot as plt

#화살표를 그리는 함수
def arrow(start, size, color):
    plt.quiver(start[0], start[1], size[0], size[1], 
               angles="xy", scale_units="xy", scale=1, color=color)
a= np.array([1,3]) #변환 전의 벡터
A= np.array([[1,-1],[2,-1]])
print(a.shape, A.shape)
b=np.dot(A,a)

print("변환 전의 벡터(a): ", a)
print("변환 후의 벡터(b): ", b)
plt.xlim(-3,3)
plt.ylim(-3,3)
plt.grid()
s= np.array([0,0])
arrow(s,a, color="black")
arrow(s,b, color="blue")
plt.gca().set_aspect("equal")
plt.show()
```
